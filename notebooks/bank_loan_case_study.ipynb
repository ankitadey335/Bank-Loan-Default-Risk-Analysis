```python
def uni_numerical_analysis(dataframe, column):
    sns.set(style="darkgrid")
    plt.figure(figsize=(18,10))

    plt.subplot(1,2,1)
    sns.boxplot(data=dataframe, x=column, orient='v').set(title='boxplot')
    plt.subplot(1,2,2)
    sns.histplot(dataframe[column], kde=True).set(title="histplot")

    plt.show()
```

```python
import warnings 
warnings.filterwarnings('ignore')
```

```python
for i in numerical_columns:
    uni_numerical_analysis(df_app, i)
```

```python
def uni_categorical_analysis(dataframe, column):
    plt.figure(figsize=(16, 10))
    sns.barplot(x=dataframe[column].value_counts(), y=dataframe[column].value_counts().index)
    plt.title(f" frequency of {column}", fontsize=15)
    plt.show()
```

```python
for i in categorical_columns:
    uni_categorical_analysis(df_app, i)
```

## 1. Which income groups are more likely to default?

```python

income_bins = pd.qcut(df_balanced['AMT_INCOME_TOTAL'], q=5)

income_default = df_balanced.groupby(income_bins)['TARGET'].mean().reset_index()

plt.figure(figsize=(12,5))
sns.barplot(x=income_default['AMT_INCOME_TOTAL'].astype(str),
            y=income_default['TARGET'])

plt.xticks(rotation=30)
plt.title('Default Rate by Income Segment')
plt.ylabel('Average Default Rate')
plt.show()
```


### Business Insight
Customers in lower income brackets showed a noticeably higher default tendency.  
This indicates repayment pressure becomes stronger as income decreases.

The bank should apply stricter credit checks for low-income applicants while offering flexible EMI structures.

```python

plt.figure(figsize=(12,5))

sns.boxplot(
    x='TARGET',
    y='DAYS_EMPLOYED',
    data=df_balanced
)

plt.title('Employment Duration vs Default')
plt.show()
```


### Business Insight
Customers with shorter employment duration were more likely to default.

Employment stability can be considered an important risk assessment parameter.

# KPI Summary

## Key Business KPIs

| KPI | Description |
|---|---|
| Total Customers | Total loan applicants in dataset |
| Default Rate | Percentage of customers who defaulted |
| Average Loan Amount | Average credit amount sanctioned |
| Average Income | Mean customer income |
| High Risk Segment | Income + employment based risky customers |
| Repeat Loan Customers | Customers with multiple previous applications |
| Approval Efficiency | Ratio of approved vs rejected applications |
| EMI Burden Ratio | Loan annuity relative to customer income |

These KPIs help management monitor portfolio health and lending quality.


```python

total_customers = df_app.shape[0]

default_rate = round((df_app['TARGET'].mean()) * 100,2)

avg_loan = round(df_app['AMT_CREDIT'].mean(),2)

avg_income = round(df_app['AMT_INCOME_TOTAL'].mean(),2)

high_risk = df_app[(df_app['AMT_INCOME_TOTAL'] < df_app['AMT_INCOME_TOTAL'].median()) & (df_app['DAYS_EMPLOYED'] < 2000)].shape[0]

repeat_customers = merged_df['SK_ID_CURR'].nunique()

approval_efficiency = round((merged_df[merged_df['NAME_CONTRACT_STATUS'] == 'Approved'].shape[0]/merged_df.shape[0]) * 100,2)

emi_burden = round((df_app['AMT_ANNUITY'] /df_app['AMT_INCOME_TOTAL']).mean() * 100,2)


kpi_names = [
    'Total Customers',
    'Default Rate (%)',
    'Average Loan Amount',
    'Average Income',
    'High Risk Customers',
    'Repeat Loan Customers',
    'Approval Efficiency (%)',
    'EMI Burden Ratio (%)'
]

kpi_values = [
    total_customers,
    default_rate,
    avg_loan,
    avg_income,
    high_risk,
    repeat_customers,
    approval_efficiency,
    emi_burden
]

kpi_df = pd.DataFrame({'KPI': kpi_names,'Value': kpi_values})

kpi_df


plt.figure(figsize=(16,8))

bars = plt.bar(kpi_df['KPI'],kpi_df['Value'])

plt.xticks(rotation=20, fontsize=11)

plt.title('Bank Loan Portfolio KPI Dashboard',fontsize=22,fontweight='bold')

plt.ylabel('KPI Values', fontsize=14)

for bar in bars:
    yval = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,yval,round(yval,2),ha='center',va='bottom',fontsize=10,fontweight='bold')

plt.tight_layout()

plt.show()
```
